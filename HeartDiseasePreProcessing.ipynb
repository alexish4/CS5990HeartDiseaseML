```python
# Get lists of numeric and categorical columns
numeric_cols = df.select_dtypes(include=['float64', 'int64']).columns.tolist()
categorical_cols = df.select_dtypes(include=['object']).columns.tolist()

# Let's print them to double-check
print("Numeric Columns:", numeric_cols)
print("Categorical Columns:", categorical_cols)
```

```python
# Fill missing numerical columns with their median value
for col in numeric_cols:
    if df[col].isnull().sum() > 0:
        median_value = df[col].median()
        df[col].fillna(median_value, inplace=True)
```

```python
# Fill missing categorical columns with their mode (most frequent value)
for col in categorical_cols:
    if col != 'Alcohol Consumption':  # Skip Alcohol Consumption for now
        if df[col].isnull().sum() > 0:
            mode_value = df[col].mode()[0]
            df[col].fillna(mode_value, inplace=True)
```

```python
from sklearn.preprocessing import LabelEncoder

# Initialize the encoder
le = LabelEncoder()

# Apply label encoding to all categorical columns
for col in categorical_cols:
    df[col] = le.fit_transform(df[col].astype(str))
```

```python
from sklearn.feature_selection import f_classif

# Perform F-test
f_values, p_values = f_classif(df.drop('Heart Disease Status', axis = 1), df['Heart Disease Status'])


# # Print the name of each feature along with corresponding F-value and p-value
# for feature, f_val, p_val in zip(X.columns, f_values, p_values):
#     print(f"Feature: {feature}\nF-value: {f_val}\nP-value: {p_val}\n")
```

```python
# Import necessary libraries
from scipy.stats import kurtosis

# Define a function to check if a variable needs exponential smoothing
def needs_smoothing(series, threshold=3):
    # Calculate the kurtosis of the series
    kurt = kurtosis(series)
    # If the kurtosis is greater than the threshold, the series needs smoothing
    return kurt > threshold

# Apply the function to each column in the DataFrame
smoothing_needed = df.apply(needs_smoothing)

# Print the columns that need smoothing
print("Columns that need smoothing:", smoothing_needed[smoothing_needed].index.tolist())

# Apply exponential smoothing to the columns that need it
for col in smoothing_needed[smoothing_needed].index:
    df[col] = np.log1p(df[col])
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from imblearn.over_sampling import SMOTE
import pandas as pd
import joblib

# Separate features and target
X = df.drop('Heart Disease Status', axis=1)
y = df['Heart Disease Status']

# Step 1: Split before SMOTE and scaling
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# Step 2: Apply SMOTE on training data only
smote = SMOTE(random_state=42)
X_train_resampled, y_train_resampled = smote.fit_resample(X_train, y_train)

# Optional: Feature selection can be done here using X_train_resampled and y_train_resampled

# Step 3: Scale both sets using scaler fitted on training data
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_resampled)
X_test_scaled = scaler.transform(X_test)

# Step 4: Save to files
joblib.dump((X_train_scaled, y_train_resampled), 'X_train_y_train.pkl')
joblib.dump((X_test_scaled, y_test), 'X_test_y_test.pkl')
```

```python
from sklearn.feature_selection import f_classif

# Perform F-test
f_values, p_values = f_classif(X, y)

# Print the name of each feature along with corresponding F-value and p-value
#for feature, f_val, p_val in zip(X.columns, f_values, p_values):
    #print(f"Feature: {feature}\nF-value: {f_val}\nP-value: {p_val}\n")
# Create a DataFrame to hold the feature names, F-values, and p-values
f_test_results = pd.DataFrame({'Feature': X.columns, 'F-value': f_values, 'P-value': p_values})

# Identify the features with F-value less than 1 and reasonable p-value 
drop_features = f_test_results[(f_test_results['F-value'] < 1) & (f_test_results['P-value'] < 0.1)]['Feature']
print(drop_features)
# Drop these features from the DataFrame
X = X.drop(columns=drop_features)


#all vars pass the f-test, tried setting stricter tresholds but worse result on preds
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Initialize the RandomForestClassifier
rf = RandomForestClassifier(random_state=42)

# Fit the model on the training data
rf.fit(X_train, y_train)

# Predict the target for the test data
y_pred = rf.predict(X_test)

# Calculate the accuracy score
accuracy = accuracy_score(y_test, y_pred)

# Print the accuracy score
print("Accuracy:", accuracy)

# Print the confusion matrix
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))

# Print the classification report
print("Classification Report:\n", classification_report(y_test, y_pred))
```
